```python
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
from torch.utils.data import DataLoader
import torchvision
import torchvision.transforms as transforms
from torchvision import models
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
import seaborn as sns
from tqdm import tqdm
import os
```

```python
# Set device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# CIFAR-10 classes
classes = ['airplane', 'automobile', 'bird', 'cat', 'deer', 'dog', 'frog', 'horse', 'ship', 'truck']

# 1. AUTOENCODER ARCHITECTURE
class Autoencoder(nn.Module):
    def __init__(self, latent_dim=128):
        super(Autoencoder, self).__init__()

        # Encoder
        self.encoder = nn.Sequential(
            nn.Conv2d(3, 32, 3, stride=2, padding=1),  # 32x16x16
            nn.ReLU(True),
            nn.Conv2d(32, 64, 3, stride=2, padding=1), # 64x8x8
            nn.ReLU(True),
            nn.Conv2d(64, 128, 3, stride=2, padding=1), # 128x4x4
            nn.ReLU(True),
            nn.Flatten(),
            nn.Linear(128 * 4 * 4, latent_dim)
        )

        # Decoder
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 128 * 4 * 4),
            nn.ReLU(True),
            nn.Unflatten(1, (128, 4, 4)),
            nn.ConvTranspose2d(128, 64, 3, stride=2, padding=1, output_padding=1), # 64x8x8
            nn.ReLU(True),
            nn.ConvTranspose2d(64, 32, 3, stride=2, padding=1, output_padding=1),  # 32x16x16
            nn.ReLU(True),
            nn.ConvTranspose2d(32, 3, 3, stride=2, padding=1, output_padding=1),   # 3x32x32
            nn.Sigmoid()
        )

    def forward(self, x):
        encoded = self.encoder(x)
        decoded = self.decoder(encoded)
        return decoded, encoded

# 2. DENOISING AUTOENCODER
class DenoisingAutoencoder(nn.Module):
    def __init__(self, latent_dim=128):
        super(DenoisingAutoencoder, self).__init__()

        # Encoder
        self.encoder = nn.Sequential(
            nn.Conv2d(3, 64, 3, stride=2, padding=1),  # 64x16x16
            nn.ReLU(True),
            nn.Conv2d(64, 128, 3, stride=2, padding=1), # 128x8x8
            nn.ReLU(True),
            nn.Conv2d(128, 256, 3, stride=2, padding=1), # 256x4x4
            nn.ReLU(True),
            nn.Flatten(),
            nn.Linear(256 * 4 * 4, latent_dim)
        )

        # Decoder
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 256 * 4 * 4),
            nn.ReLU(True),
            nn.Unflatten(1, (256, 4, 4)),
            nn.ConvTranspose2d(256, 128, 3, stride=2, padding=1, output_padding=1), # 128x8x8
            nn.ReLU(True),
            nn.ConvTranspose2d(128, 64, 3, stride=2, padding=1, output_padding=1),  # 64x16x16
            nn.ReLU(True),
            nn.ConvTranspose2d(64, 3, 3, stride=2, padding=1, output_padding=1),   # 3x32x32
            nn.Sigmoid()
        )

    def forward(self, x):
        encoded = self.encoder(x)
        decoded = self.decoder(encoded)
        return decoded, encoded

# 3. CNN CLASSIFIER
class CNNClassifier(nn.Module):
    def __init__(self, num_classes=10):
        super(CNNClassifier, self).__init__()

        self.features = nn.Sequential(
            # First block
            nn.Conv2d(3, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.Conv2d(64, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2, 2),

            # Second block
            nn.Conv2d(64, 128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.Conv2d(128, 128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2, 2),

            # Third block
            nn.Conv2d(128, 256, 3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),
            nn.Conv2d(256, 256, 3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2, 2),
        )

        self.classifier = nn.Sequential(
            nn.Dropout(0.5),
            nn.Linear(256 * 4 * 4, 512),
            nn.ReLU(inplace=True),
            nn.Dropout(0.5),
            nn.Linear(512, num_classes)
        )

    def forward(self, x):
        x = self.features(x)
        x = x.view(x.size(0), -1)
        x = self.classifier(x)
        return x

# 4. DATA LOADING AND PREPROCESSING
def get_data_loaders():
    # Basic transform for autoencoder
    transform_basic = transforms.Compose([
        transforms.ToTensor(),
    ])

    # Transform with data augmentation for CNN
    transform_augmented = transforms.Compose([
        transforms.RandomHorizontalFlip(p=0.5),
        transforms.RandomRotation(10),
        transforms.RandomResizedCrop(32, scale=(0.8, 1.0)),
        transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.1),
        transforms.ToTensor(),
        transforms.Normalize((0.485, 0.456, 0.406), (0.229, 0.224, 0.225))
    ])

    # Transform for testing (no augmentation)
    transform_test = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize((0.485, 0.456, 0.406), (0.229, 0.224, 0.225))
    ])

    # Datasets
    train_dataset_basic = torchvision.datasets.CIFAR10(root='./data', train=True,
                                                      download=True, transform=transform_basic)
    train_dataset_augmented = torchvision.datasets.CIFAR10(root='./data', train=True,
                                                          download=True, transform=transform_augmented)
    test_dataset = torchvision.datasets.CIFAR10(root='./data', train=False,
                                               download=True, transform=transform_test)

    # Data loaders
    train_loader_basic = DataLoader(train_dataset_basic, batch_size=128, shuffle=True)
    train_loader_augmented = DataLoader(train_dataset_augmented, batch_size=128, shuffle=True)
    test_loader = DataLoader(test_dataset, batch_size=128, shuffle=False)

    return train_loader_basic, train_loader_augmented, test_loader

# 5. TRAINING FUNCTIONS
def train_autoencoder(model, train_loader, num_epochs=20):
    criterion = nn.MSELoss()
    optimizer = optim.Adam(model.parameters(), lr=0.001)

    model.train()
    for epoch in range(num_epochs):
        total_loss = 0
        for batch_idx, (data, _) in enumerate(tqdm(train_loader, desc=f'Epoch {epoch+1}/{num_epochs}')):
            data = data.to(device)

            optimizer.zero_grad()
            reconstructed, _ = model(data)
            loss = criterion(reconstructed, data)
            loss.backward()
            optimizer.step()

            total_loss += loss.item()

        print(f'Epoch {epoch+1}/{num_epochs}, Loss: {total_loss/len(train_loader):.4f}')

def train_denoising_autoencoder(model, train_loader, num_epochs=20, noise_factor=0.3):
    criterion = nn.MSELoss()
    optimizer = optim.Adam(model.parameters(), lr=0.001)

    model.train()
    for epoch in range(num_epochs):
        total_loss = 0
        for batch_idx, (data, _) in enumerate(tqdm(train_loader, desc=f'Epoch {epoch+1}/{num_epochs}')):
            data = data.to(device)

            # Add noise to input
            noise = torch.randn_like(data) * noise_factor
            noisy_data = torch.clamp(data + noise, 0, 1)

            optimizer.zero_grad()
            reconstructed, _ = model(noisy_data)
            loss = criterion(reconstructed, data)  # Reconstruct original, not noisy
            loss.backward()
            optimizer.step()

            total_loss += loss.item()

        print(f'Epoch {epoch+1}/{num_epochs}, Loss: {total_loss/len(train_loader):.4f}')

def train_cnn_classifier(model, train_loader, test_loader, num_epochs=50):
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=0.001)
    scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=20, gamma=0.1)

    best_acc = 0.0

    for epoch in range(num_epochs):
        # Training
        model.train()
        train_loss = 0
        train_correct = 0
        train_total = 0

        for batch_idx, (data, target) in enumerate(tqdm(train_loader, desc=f'Epoch {epoch+1}/{num_epochs}')):
            data, target = data.to(device), target.to(device)

            optimizer.zero_grad()
            output = model(data)
            loss = criterion(output, target)
            loss.backward()
            optimizer.step()

            train_loss += loss.item()
            _, predicted = output.max(1)
            train_total += target.size(0)
            train_correct += predicted.eq(target).sum().item()

        # Testing
        model.eval()
        test_loss = 0
        test_correct = 0
        test_total = 0

        with torch.no_grad():
            for data, target in test_loader:
                data, target = data.to(device), target.to(device)
                output = model(data)
                test_loss += criterion(output, target).item()
                _, predicted = output.max(1)
                test_total += target.size(0)
                test_correct += predicted.eq(target).sum().item()

        train_acc = 100. * train_correct / train_total
        test_acc = 100. * test_correct / test_total

        print(f'Epoch {epoch+1}/{num_epochs}:')
        print(f'  Train Loss: {train_loss/len(train_loader):.4f}, Train Acc: {train_acc:.2f}%')
        print(f'  Test Loss: {test_loss/len(test_loader):.4f}, Test Acc: {test_acc:.2f}%')

        if test_acc > best_acc:
            best_acc = test_acc
            torch.save(model.state_dict(), 'best_cnn_model.pth')

        scheduler.step()

    return best_acc

# 6. FEATURE EXTRACTION AND VISUALIZATION
def extract_features(model, data_loader, model_type='autoencoder'):
    model.eval()
    features = []
    labels = []

    with torch.no_grad():
        for data, target in data_loader:
            data = data.to(device)

            if model_type == 'autoencoder':
                _, encoded = model(data)
                features.append(encoded.cpu().numpy())
            elif model_type == 'cnn':
                # Extract features from the last conv layer
                x = model.features(data)
                x = x.view(x.size(0), -1)
                features.append(x.cpu().numpy())

            labels.append(target.numpy())

    return np.concatenate(features), np.concatenate(labels)

def get_pretrained_cnn_features(data_loader):
    # Use ResNet18 pretrained on ImageNet
    resnet = models.resnet18(pretrained=True)
    resnet.eval()
    resnet = resnet.to(device)

    # Remove the final classification layer
    feature_extractor = nn.Sequential(*list(resnet.children())[:-1])

    features = []
    labels = []

    with torch.no_grad():
        for data, target in data_loader:
            data = data.to(device)
            # ResNet expects normalized data
            normalize = transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
            data = normalize(data)

            feat = feature_extractor(data)
            feat = feat.view(feat.size(0), -1)
            features.append(feat.cpu().numpy())
            labels.append(target.numpy())

    return np.concatenate(features), np.concatenate(labels)

def visualize_features(features, labels, title, method='tsne'):
    plt.figure(figsize=(12, 8))

    if method == 'pca':
        reducer = PCA(n_components=2)
        features_2d = reducer.fit_transform(features)
        explained_var = reducer.explained_variance_ratio_
        subtitle = f'PCA (explained variance: {explained_var[0]:.2f}, {explained_var[1]:.2f})'
    elif method == 'tsne':
        reducer = TSNE(n_components=2, random_state=42, perplexity=30)
        features_2d = reducer.fit_transform(features)
        subtitle = 't-SNE'

    # Create scatter plot
    scatter = plt.scatter(features_2d[:, 0], features_2d[:, 1], c=labels, cmap='tab10', s=20, alpha=0.6)
    plt.colorbar(scatter, ticks=range(10), label='Class')
    plt.clim(0, 9)

    # Add class names to colorbar
    cbar = plt.colorbar(scatter, ticks=range(10))
    cbar.set_ticklabels(classes)

    plt.title(f'{title}\n{subtitle}')
    plt.xlabel('Component 1')
    plt.ylabel('Component 2')
    plt.tight_layout()
    plt.show()

# 7. MAIN EXECUTION
def main():
    # Get data loaders
    train_loader_basic, train_loader_augmented, test_loader = get_data_loaders()

    print("=" * 60)
    print("1. TRAINING AUTOENCODER FOR 2D FEATURE GENERATION")
    print("=" * 60)

    # Train autoencoder
    autoencoder = Autoencoder(latent_dim=128).to(device)
    train_autoencoder(autoencoder, train_loader_basic, num_epochs=20)

    # Extract autoencoder features
    ae_features, ae_labels = extract_features(autoencoder, test_loader, 'autoencoder')

    print("\n" + "=" * 60)
    print("2. COMPARING AUTOENCODER VS PRETRAINED CNN FEATURES")
    print("=" * 60)

    # Get pretrained CNN features
    cnn_features, cnn_labels = get_pretrained_cnn_features(test_loader)

    # Visualize autoencoder features
    print("Visualizing Autoencoder Features with t-SNE...")
    visualize_features(ae_features, ae_labels, "Autoencoder Features", 'tsne')

    print("Visualizing Autoencoder Features with PCA...")
    visualize_features(ae_features, ae_labels, "Autoencoder Features", 'pca')

    # Visualize pretrained CNN features
    print("Visualizing Pretrained CNN Features with t-SNE...")
    visualize_features(cnn_features, cnn_labels, "Pretrained ResNet18 Features", 'tsne')

    print("Visualizing Pretrained CNN Features with PCA...")
    visualize_features(cnn_features, cnn_labels, "Pretrained ResNet18 Features", 'pca')

    print("\n" + "=" * 60)
    print("3. TRAINING DENOISING AUTOENCODER")
    print("=" * 60)

    # Train denoising autoencoder
    denoising_ae = DenoisingAutoencoder(latent_dim=128).to(device)
    train_denoising_autoencoder(denoising_ae, train_loader_basic, num_epochs=20)

    # Visualize denoising results
    denoising_ae.eval()
    with torch.no_grad():
        sample_data, _ = next(iter(test_loader))
        sample_data = sample_data[:8].to(device)

        # Add noise
        noise = torch.randn_like(sample_data) * 0.3
        noisy_data = torch.clamp(sample_data + noise, 0, 1)

        # Reconstruct
        reconstructed, _ = denoising_ae(noisy_data)

        # Plot results
        fig, axes = plt.subplots(3, 8, figsize=(16, 6))
        for i in range(8):
            # Original
            axes[0, i].imshow(sample_data[i].cpu().permute(1, 2, 0))
            axes[0, i].set_title('Original')
            axes[0, i].axis('off')

            # Noisy
            axes[1, i].imshow(noisy_data[i].cpu().permute(1, 2, 0))
            axes[1, i].set_title('Noisy')
            axes[1, i].axis('off')

            # Reconstructed
            axes[2, i].imshow(reconstructed[i].cpu().permute(1, 2, 0))
            axes[2, i].set_title('Denoised')
            axes[2, i].axis('off')

        plt.tight_layout()
        plt.show()

    print("\n" + "=" * 60)
    print("4. TRAINING CNN WITHOUT DATA AUGMENTATION")
    print("=" * 60)

    # Train CNN without augmentation
    cnn_no_aug = CNNClassifier().to(device)
    acc_no_aug = train_cnn_classifier(cnn_no_aug, train_loader_basic, test_loader, num_epochs=50)

    print("\n" + "=" * 60)
    print("5. TRAINING CNN WITH DATA AUGMENTATION")
    print("=" * 60)

    # Train CNN with augmentation
    cnn_with_aug = CNNClassifier().to(device)
    acc_with_aug = train_cnn_classifier(cnn_with_aug, train_loader_augmented, test_loader, num_epochs=50)

    print("\n" + "=" * 60)
    print("FINAL RESULTS SUMMARY")
    print("=" * 60)
    print(f"CNN without augmentation - Best Test Accuracy: {acc_no_aug:.2f}%")
    print(f"CNN with augmentation - Best Test Accuracy: {acc_with_aug:.2f}%")
    print(f"Improvement with augmentation: {acc_with_aug - acc_no_aug:.2f}%")

if __name__ == "__main__":
    main()
```
